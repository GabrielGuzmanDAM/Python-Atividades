```python
# Ejercicio 1
#Pedir al usuario una nota numérica entera mediante un input y diga si la calificación
#es Suspenso, Aprobado, Notable, Sobresaliente, o No válida (en caso de que la
#entrada sea diferente a la esperada). Realizar una versión con IF y otra con MATCH.

try:
    nota = int(input("Introduce una nota entera (0-10): "))
except ValueError:
    nota = None

# Versión con if
if nota is None or not 0 <= nota <= 10:
    print("IF: No válida")
elif nota < 5:
    print("IF: Suspenso")
elif nota < 7:
    print("IF: Aprobado")
elif nota < 9:
    print("IF: Notable")
else:
    print("IF: Sobresaliente")

# Versión con match 
match nota:
    case n if n is None or not 0 <= n <= 10:
        print("MATCH: No válida")
    case n if n < 5:
        print("MATCH: Suspenso")
    case n if n < 7:
        print("MATCH: Aprobado")
    case n if n < 9:
        print("MATCH: Notable")
    case _:
        print("MATCH: Sobresaliente")
```

```python
# Ejercicio 2
#Pedir al usuario dos valores por pantalla: el precio de un producto (float) y el tipo de
#IVA (General, Reducido, Superreducido). Calcular el precio final del producto fruto
#de sumarle el IVA en función de su tipo. Realizar una versión con IF y otra con MATCH.

producto = None;
tipo_iva = None;

while producto == None:
    try:
        producto = float(input("Dame el precio de un producto: "))
    except ValueError:
        producto = None

tipo_iva_lista = ["General", "Reducido", "Superreducido"]

while tipo_iva == None:
    try:
        tipo_iva = str(input("Dame el tipo de IVA (General, Reducido, Superreducido): "))

        if tipo_iva not in tipo_iva_lista:
            print("Tipo de IVA no válido. Por favor, introduce uno de los siguientes: General, Reducido, Superreducido.")
            tipo_iva = None
    except ValueError:
        tipo_iva = None

if tipo_iva == "General":
    precio_final = producto * 1.21
elif tipo_iva == "Reducido":
    precio_final = producto * 1.10
elif tipo_iva == "Superreducido":
    precio_final = producto * 1.04

print(f"Precio del producto: {producto}€")
print(f"Tipo de IVA: {tipo_iva}")
print(f"Precio final del producto: {precio_final:.2f}€")
```

```python
# Ejercicio 4
#A partir de 60 mm de lluvia acumulados en 12 horas se declara una alerta amarilla, y
#a partir de 120 mm, una alerta roja. Pedir al usuario los milímetros de lluvia
#acumulados y mostrar por pantalla si No hay alerta, Hay alerta amarilla o Hay
#alerta roja. Realizar una versión con IF y otra con MATCH.

milimetros_lluvia = None;

while milimetros_lluvia == None:
    try:
        milimetros_lluvia = int(input("Introduce el milimetros de lluvia: "))
    except ValueError:
        milimetros_lluvia = None

# Versión con if
if milimetros_lluvia >= 120:
    print("Alarma: ROJA")
elif milimetros_lluvia >= 60:
    print("Alarma: AMARILLA")
else:
    print("No hay alarma")

# Versión con MATCH
match milimetros_lluvia:

    case n if n > 120:
        print("Alarma: ROJA")

    case n if n > 60:
        print("Alarma: AMARILLA")

    case _:
        print("No hay alarma")
    


```

```python
# Ejercicio 5
#Pedir al usuario dos números enteros y mostrar los números pares dentro de dicho
#intervalo. Si el primer número es mayor que el primero se mostrará por pantalla un
#mensaje de error. Realizar una versión con FOR y otra con WHILE.

numero1 = None;
numero2 = None;

while numero1 == None:
    try:
        numero1 = int(input("Introdusca el primer numero entero:"))
    except ValueError:
            numero1 = None

while numero2 == None:
    try:
        numero2 = int(input("Introdusca el segundo numero entero:"))
    except ValueError:
            numero2 = None

if numero1 <= numero2:
    for i in range(numero1, numero2 + 1):
        if i % 2 == 0:
            print(f"FOR: {i} es par")

    while numero1 <= numero2:
        if numero1 % 2 == 0:
            print(f"WHILE: {numero1} es par")
        numero1 += 1
elif numero1 > numero2:
        print("Error: El primer número debe ser menor o igual al segundo número.")
    
        
```

```python
# Ejercicio 6
#Supongamos que la contraseña para acceder es “12345”. Pedir al usuario la
#contraseña por pantalla y, si es la correcta, mostrar un mensaje de bienvenida. Si la
#contraseña introducida es incorrecta, volver a pedirla hasta que se introduzca
#correctamente.

contrasena = ""

contrasena_correcta = "12345"

while contrasena != contrasena_correcta:
    
    contrasena = input("Introduce la contraseña: ")

    if contrasena == contrasena_correcta:
        print("Bienvenido!")
    else:
        print("Contraseña incorrecta. Inténtalo de nuevo.")
```

```python
# Ejercicio 9
# Pedir al usuario un número entero y mostrar por pantalla si el número es primo o no.

numero = None;

while numero == None:
    try:
        numero = int(input("Introduce un número entero: "))
    except ValueError:
        numero = None

if numero < 2:
    print(f"{numero} no es primo.")
else:
    es_primo = True
    for i in range(2, int(numero ** 0.5) + 1):
        if numero % i == 0:
            es_primo = False
            break
    if es_primo:
        print(f"{numero} es primo.")
    else:
        print(f"{numero} no es primo.")
```

```python
# Ejercicio 10
#Pedir al usuario dos valores enteros y mostrar por pantalla todos los números primos
#dentro de ese rango. Si el primer número es mayor que el primero se mostrará por
#pantalla un mensaje de error.

numero1 = None;
numero2 = None;

while numero1 == None:
    try:
        numero1 = int(input("Introduce el primer número entero: "))
    except ValueError:
        numero1 = None

while numero2 == None:
    try:
        numero2 = int(input("Introduce el segundo número entero: "))
    except ValueError:
        numero2 = None

if numero1 > numero2:
    print("Error: El primer número debe ser menor o igual al segundo número.")
else:
    print(f"Los números primos en el rango [{numero1}, {numero2}] son:")
    for num in range(numero1, numero2 + 1):
        if num < 2:
            continue
        es_primo = True
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                es_primo = False
                break
        if es_primo:
            print(num)
```
