```python
# Ejercicio 1,
# Lista de Pokemons
from datos import pokemons

# Muestra la media de altura de todos los pokémons.
alturaSumas = 0
alturaMedia = 0

for pokemon in pokemons:
    alturaSumas = alturaSumas + pokemon['altura_m'];

alturaMedia = alturaSumas / len(pokemons)

print(f"La altura media de los pokemons es: {alturaMedia}")
```

```python
# Ejercicio 2
# Lista de Pokemons
from datos import pokemons

# Muestra el nombre y la altura de todos los pokémons que midan menos que la media.

for pokemon in pokemons:
    if alturaMedia > pokemon['altura_m']:
        print(f"{pokemon['nombre']} {pokemon['altura_m']}")
```

```python
# Ejercicio 3
# Lista de Pokemons
from datos import pokemons

# Muestra en una sóla línea el nombre y el peso del pokémon con más peso.
for pokemon in pokemons:
    if pokemon["peso_kg"] == max(pokemon["peso_kg"] for pokemon in pokemons):
        print(f"{pokemon['nombre']} {pokemon['peso_kg']}")
```

```python
# Ejercicio 4
# Lista de Pokemons
from datos import pokemons

# Muestra el nombre y los tipos de los pokémons que sean de tipo agua.

for pokemon in pokemons:
    for tipo in pokemon['tipos']:
        if "Agua" == tipo:
            print(f"{pokemon['nombre']} {tipo}")
```

```python
# Ejercicio 5
# Lista de Pokemons
from datos import pokemons

# Muestra los pokémons que tengan un tipo que acabe en “a”.

for pokemon in pokemons:
    for tipo in pokemon['tipos']:
        if "a" == tipo[-1]:
            print(f"{pokemon['nombre']} {tipo}")
```
